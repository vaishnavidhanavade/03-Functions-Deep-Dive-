```python
def decor(func):
    def wrapper(x,y):
        if x is str or y is str:
            x = int(x)
            y = int(y)
            print(x/y)
        func(x,y)
    return wrapper



@decor
def devision(x,y):
    print(x/y)
    
devision("4","4")
```

```python
# Convert values into int from string
def my_devi_decor(func):
    def wrapper(x,y):
        if isinstance(x, str):
            x = int(x)
        if isinstance(y, str):
            y = int(y)
            func(x,y)
        if y>x:
            x,y = y,x
            print("Value Swapped")
        return func(x,y)
    return wrapper

@my_devi_decor
def OriginalFun(x,y):
    print(x/y)
OriginalFun('5',10)
```
